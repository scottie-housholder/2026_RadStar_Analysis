# Load Libraries and Data

```python
import pandas as pd
import geopandas as gpd
import numpy as np
import seaborn as sns

import matplotlib.pyplot as plt

from shapely.geometry import Point


#Load the cleaned data
clean_rad = pd.read_excel("radstar_temps.xlsx")
```

#### Create geodataframe for geopandas

```python
url = "https://naciscdn.org/naturalearth/110m/cultural/ne_110m_admin_0_countries.zip"
world = gpd.read_file(url)

geometry = [Point(xy) for xy in zip(clean_rad["lon"], clean_rad["lat"])]
geo_rad = gpd.GeoDataFrame(clean_rad, geometry=geometry, crs="EPSG:4326")
```

### Define a constant colormap to use for all visuals

Default: `managua` 

`managua` is very good for highlighting all points on the graph.  
Change to `Reds` if you would rather only easily see extreme high values. All low values become very hard to see on the light background

```python
COLOR = "managua"
```

# Quick look at radioIDs

```python
geo_rad.recent_radio.value_counts().plot(kind="bar")
```

Going back to the original data for a second, we had 4 different radio IDs. This is how they shake out: 

RadioID | Entries | Nulls
---|---|---
10006 | 27111 | 27111
10162 | 14658 | 14658
10012 |  2798 | 2189
10009 |   675 | 189

(This data came from my cleaning file)

We can see here that radioIDs `10006` and `10162` are both entirely nulls.

There were 84 non-null temperature values in the time window of our data. 81 of them were ID `10009`, and 3 of them were ID 10012, but all three of ID `10012` had a value of ID `10009` take place before another RadStar packet took place. This is how the above graph occured. All of our temerature data (with the way I backfilled the data) came from the radio with ID `10009`.

```python
fig, ax = plt.subplots(figsize=(14, 6))

world.plot(ax=ax, color="lightgrey")
geo_rad.plot(ax=ax,
            markersize=10,
            alpha=0.5,
            column="recent_temp", 
            cmap="bwr",
            legend=True,
            legend_kwds = {"label": "Temperature (Celsius)"}
            )

ax.set_aspect("equal", adjustable="box")
plt.title("Temperature of the satellite")
plt.show()
```

With the way I backfilled the data, it appears that satellite temperature fluctuation was not based on geographic location
